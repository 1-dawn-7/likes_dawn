### Задача 1: сломанный признак (а может и не один)
__(1 балл)__

Загрузите таблицу data.csv.

Найдите в данных сломанный признак (он не соответствует описанию) и исправьте его.

```python
import pandas as pd
import numpy as np

data = pd.read_csv('data.csv')
df_clean = data.copy()

if (df_clean['age'] > 30).any():
    df_clean.loc[df_clean['age'] > 30, 'age'] = 2006 - df_clean.loc[df_clean['age'] > 30, 'age']

if (df_clean['traveltime'] > 4).any():
    def fix_traveltime(val):
        if val <= 4: return val
        if val < 15: return 1
        if 15 <= val <= 30: return 2
        if 30 < val <= 60: return 3
        return 4
    df_clean['traveltime'] = df_clean['traveltime'].apply(fix_traveltime)

if 'plans_university' in df_clean.columns:
    df_clean['plans_university'] = df_clean['plans_university'].apply(
        lambda x: 1 if str(x).strip().lower() in ['yes', '1', '1.0'] else -1
    )

num_cols = df_clean.select_dtypes(include=[np.number]).columns
for col in num_cols:
    if col != 'plans_university' and (df_clean[col] < 0).any():
        df_clean[col] = df_clean[col].abs()
```

### Задача 2: пропуски в данных 
__(1 балл)__

Проверьте, есть ли в данных пропуски (значения NaN). Замените все пропущенные значения на среднее значение этого признака по столбцу.

*Hint: изучите в pandas функции loc, isnull, а также передачу булевых массивов в качестве индексов.*

```python
df_clean = df_clean.fillna(df_clean.mean())
```

### Задача 3: нормализация данных
__(1 балл)__

Нормализуйте данные любым способом, в современных библиотеках нормализация часто встроена внутрь модели, поэтому тут разницы может и не быть в качестве, но сделать надо

```python
X_scaled = (df_clean - df_clean.mean()) / df_clean.std()
```

### Задача 4: кросс-валидация для исходных данных
__(1 балл)__

Загрузите файл scores.csv и протестируйте, как линейная регрессия предсказывает ответ сейчас (с помощью кросс-валидации).

Кроссвалидацию сделайте по 4 разбивкам. Выведите качество в каждом их разбиений.

*Hint: воспользуйтесь sklearn.linear_model и sklearn.model_selection.*

```python
from sklearn import linear_model
from sklearn.model_selection import cross_val_score

scores = pd.read_csv('scores.csv', header=None).squeeze()

model = linear_model.LinearRegression()
cv_scores_initial = cross_val_score(model, X_scaled, scores, cv=4, scoring='neg_mean_squared_error')

print("MSE на 4 разбиениях:", -cv_scores_initial)
print("Среднее MSE:", -cv_scores_initial.mean())
```

### Задача 5: полные данные
__(2 балла)__

Воспользуйтесь файлами attendance.csv и school_support.txt для того, чтобы добавить новые признаки в данные. Желательно по максимуму использовать возможности pandas для упрощения преобразований.

school_suport число в строке значит что i-ый школьник из исходной таблицы получал мат помощь (обратите внимание что строк в файле меньше, подумайте как правильно импортировать данные)

Добавьте данные таким образом, чтобы качество выросло

```python
support_indices = pd.read_csv('school_support.txt', header=None).squeeze().values
X_scaled['school_support'] = 0
X_scaled.loc[support_indices, 'school_support'] = 1

with open('attendance.csv', 'r') as f:
    attendance_lines = f.readlines()

attendance_counts = [line.count('+') for line in attendance_lines]
X_scaled['attendance'] = pd.Series(attendance_counts[:len(X_scaled)])

cv_scores_full = cross_val_score(model, X_scaled, scores, cv=4, scoring='neg_mean_squared_error')
print("Среднее MSE после добавления признаков:", -cv_scores_full.mean())
```
